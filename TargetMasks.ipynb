```python
viz = RasterLabelVisualizer(label_map='xBD_label_map.json')
```

```python
l = viz.plot_color_legend()
```

```python
# pre_img = "palu-tsunami_00000152_pre_disaster.png"
# post_img = "palu-tsunami_00000152_post_disaster.png"

pre_mask = "palu-tsunami_00000152_pre_disaster_b2.png"
post_mask = "palu-tsunami_00000152_post_disaster_b2.png"
```

```python
preImg = np.array(Image.open(pre_img))
preMask = np.array(Image.open(pre_mask))
preMask, buf = viz.show_label_raster(preMask, size=(20, 20))
f, ax = plt.subplots(1,2, figsize=(30,9))
ax[0].imshow(preImg)
ax[0].axis('off')
ax[1].imshow(preMask)
ax[1].axis('off')
f.tight_layout()
f.show()
```
